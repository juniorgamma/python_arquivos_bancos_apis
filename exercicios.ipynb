Crie uma lista com seus filmes favoritos;

Crie uma tupla com as datas que você assistiu esses filmes;

A partir da lista, crie um dicionário onde a chave é o nome do filme e depois adicione a data que você assistiu o filme como valor.




```python
filmes = ['O poderoso chefão', 'Um sonho de liberdade', 'Donnie Darko']
datas = '01/08/2000', '25/06/2019', '30/04/2020'

filmes_dict = {filme:data for filme, data in zip(filmes, datas)}

print(filmes)
print(datas)
print(filmes_dict)
```

```python
for chave, valor in filmes_dict.items():
    print(f'{chave}: {valor}')
```

1. Crie um programa que grave em um arquivo alunos.csv uma lista de alunos e suas notas.
2. Leia o arquivo alunos.csv e imprima apenas os alunos com nota maior ou igual a 7.0.

```python
import csv

alunos = input('Digite os nomes dos alunos separados por vírgula: ').split(',')
notas = input('Digite as notas correspondentes separadas por vírgula: ').split(',')

alunos = [aluno.strip() for aluno in alunos]
notas = [float(nota.strip()) for nota in notas]

with open ('alunos.csv', 'w', newline = '', encoding = 'utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['Aluno', 'Nota'])
    for aluno, nota in zip(alunos, notas):
        writer.writerow([aluno, nota])

with open('alunos.csv', 'r', newline = '', encoding = 'utf-8') as f:
    reader = csv.reader(f)
    for i, linha in enumerate(reader):
        if i == 0:
            print(linha)
        else:
            if float(linha[1]) >= 7:
                print(linha)
```

id (inteiro, chave primária)

nome (texto)

email (texto)

Para testar o seu banco de dados, adicione 2 usuários na tabela usuarios e depois consulte todos os registros.

```python
import sqlite3

conn = sqlite3.connect('hotelplus.db')

cursor = conn.cursor()

cursor.execute(
    '''
        CREATE TABLE IF NOT EXISTS usuarios(
            id INTEGER PRIMARY KEY,
            nome TEXT,
            email TEXT
        )
    '''
)

conn.commit()

conn.close()
```

A Loja TechMais está informatizando seu controle de estoque
O gerente quer um sistema simples em Python que registre e consulte produtos no banco de dados loja.db.

Se conecta ao banco loja.db;

Crie uma tabela produtos com os campos id, nome, preco;

Insira 3 produtos diferentes;

Liste todos os produtos cadastrados.

Dica: use funções como criar_produto(nome, preco) e listar_produtos().

```python
import sqlite3
import pandas as pd

def conectar():
    conn = sqlite3.connect('loja.db')
    cursor = conn.cursor()
    return conn, cursor

def desconectar(conn):
    conn.commit()
    conn.close()    

def criar_tabela_produtos():
    conn, cursor = conectar()
    cursor.execute(
        '''
            CREATE TABLE IF NOT EXISTS produtos(
                id INTEGER PRIMARY KEY,
                nome TEXT,
                preco FLOAT
            )
        '''
    )
    desconectar(conn)

def criar_produtos(lista_produtos):
    conn, cursor = conectar()
    cursor.executemany(
        '''
            INSERT INTO produtos(nome, preco)
            VALUES(?, ?)
        ''',
        lista_produtos
    )
    desconectar(conn)

def listar_produtos():
    conn, cursor = conectar()
    cursor.execute(
        '''
            SELECT * FROM produtos
        '''
    )
    produtos = cursor.fetchall()
    df = pd.DataFrame(produtos, columns = ['id', 'nome', 'preco'])
    desconectar(conn)
    return df

def main():
    conectar()
    criar_tabela_produtos()
    lista_produtos = [
        ('PS5', 3599.99),
        ('Macbook', 19999.99),
        ('Apple Watch', 6899.99)
    ]
    criar_produtos(lista_produtos)
    df = listar_produtos()
    display(df)

if __name__ == '__main__':
    main()
```
